```python
center = (40.399, -3.744)

m = Map(center=center, zoom=10)

circle = CircleMarker()
circle.location = center
circle.radius = 5
# circle.color = "green"
# circle.fill_color = None
circle.stroke = True
circle.color = "green"
circle.weight = 1
circle.fill = True
circle.fill_color = None
circle.fill_opacity = 1
circle.line_cap = "round"
circle.line_join = "round"

m.add(circle)

m
```
